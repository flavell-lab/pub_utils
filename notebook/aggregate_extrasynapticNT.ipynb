```python
import pickle, pandas as pd                                                                                                                                         
                                                                                                                                                                    
with open('../connectomes/candy_assembly/aggregated_extrasynapticNT/aggregated_extrasynapticNT_single.pkl', 'rb') as f:                                                
    single = pickle.load(f)                                                                                                                                         
                                                                                                                                                                    
na = pd.read_csv('../connectomes/neuroanatomy.csv')                                                                                                                    
neuron_to_figclass = dict(zip(na['neuronID'], na['figNeuronClass']))                                                                                                
classes_order = list(dict.fromkeys(na['figNeuronClass'].dropna()))                                                                                                  
                                                                                                                                                                    
result = {}
for key, df in single.items():                                                                                                                                      
    row_classes = [neuron_to_figclass[n] for n in df.index]                                                                                                       
    col_classes = [neuron_to_figclass[n] for n in df.columns]
                                                                                                                                                                    
    df_grouped = df.copy()
    df_grouped.index = row_classes                                                                                                                                  
    df_grouped.columns = col_classes                                                                                                                              
    df_grouped = df_grouped.loc[df_grouped.index.notna(), df_grouped.columns.notna()]
                                                                                                                                                                    
    df_class = df_grouped.groupby(level=0, sort=False).mean()
    df_class = df_class.T.groupby(level=0, sort=False).mean().T                                                                                                     
                                                                                                                                                                
    present = [c for c in classes_order if c in df_class.index]                                                                                                     
    df_class = df_class.loc[present, present]
    result[key] = df_class                                                                                                                                          
                                                                                                                                                                    
with open('../connectomes/candy_assembly/aggregated_extrasynapticNT/aggregated_extrasynapticNT_single_by_class.pkl', 'wb') as f:
    pickle.dump(result, f)      
```
